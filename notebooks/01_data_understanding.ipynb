## Initial Data Quality Assessment

### Missing Value Analysis

This section examines conventional missing values across the
three datasets. Coded missing or unknown values will be
investigated separately using the official STATS19 data guide.

```python
datasets = {"Collisions": collisions,"Vehicles": vehicles,"Casualties": casualties}

for name, df in datasets.items():
    print(f"\n{name} Dataset")
    print("-" * 40)

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No conventional missing values found.")
    else:
        print(missing.sort_values(ascending=False))
```


### Findings — Conventional Missing Values

- The collision dataset contains 8 missing cells across 4 geographical columns: location_easting_osgr, location_northing_osgr, longitude, and latitude.
- Each affected column contains 2 missing values.
- No conventional missing values were detected in the vehicle or casualty datasets.
- The missing geographical values may belong to the same collision records; this will be verified during further assessment.
- Conventional missing-value analysis alone is insufficient because STATS19 uses coded values for unknown, unavailable, and inapplicable information.

**Conclusion:** The datasets have very low conventional missingness. Further investigation of coded missing values is required before determining their overall data quality.

### Investigation of Coded Missing and Unknown Values

The STATS19 datasets use numerical codes to represent certain
unknown, unavailable, or inapplicable values.

This section identifies occurrences of -1 across the datasets.
The official data dictionary will be used to interpret these
codes before any cleaning decisions are made.

```python
for name, df in datasets.items():
    print(f"\n{name} Dataset")
    print("-" * 50)

    numeric_columns = df.select_dtypes(include="number")

    minus_one_counts = (numeric_columns == -1).sum()
    minus_one_counts = minus_one_counts[minus_one_counts > 0]

    if minus_one_counts.empty:
        print("No -1 values found.")
    else:
        result = pd.DataFrame({"Count": minus_one_counts, "Percentage (%)":(minus_one_counts / len(df) * 100).round(2)})
        print(result.sort_values("Count", ascending=False))
```


### Findings — Coded Missing and Unknown Values

The analysis identified widespread occurrences of `-1` across all three datasets, despite the very low conventional missing-value counts.

**Collision Dataset**
- `local_authority_district` contains `-1` in 100% of records.
- Several historical variables contain approximately 86–87% `-1` values.
- `junction_control` contains 43.82% `-1` values.
- Some road and severity-related attributes also contain coded unavailable values.

**Vehicle Dataset**
- `age_of_vehicle` contains `-1` in 100% of records.
- Historical vehicle attributes contain approximately 87% `-1` values.
- `engine_capacity_cc`, `driver_imd_decile`, and `propulsion_code` contain substantial proportions of `-1`.
- Driver age information is unavailable or specially coded in 15.36% of records.

**Casualty Dataset**
- `pedestrian_road_maintenance_worker` contains 13.65% `-1` values.
- `enhanced_casualty_severity` contains 13.60% `-1` values.
- Casualty socioeconomic and distance-banding fields contain 11.47% `-1`.
- Age and sex attributes contain smaller proportions of coded unavailable values.

**Interpretation**

The results indicate that conventional null-value detection substantially understates the data-quality challenges in STATS19.

However, `-1` occurrences cannot automatically be classified as missing values. Their meanings must be verified against the official data dictionary, particularly for historical and potentially inapplicable variables.

**Conclusion:** Column-specific code interpretation will be essential during the cleaning stage. No values have been replaced or removed during this investigation.

### Official STATS19 Data Dictionary Examination

The official STATS19 data guide provides definitions of
variables, categorical codes, and reporting conventions.

This section examines the guide to distinguish missing,
unknown, not applicable, and historical reporting values.

```python
data_guide_path = next(data_dir.glob("*.xlsx"))
excel_file = pd.ExcelFile(data_guide_path)

print("Available sheets:")
print(excel_file.sheet_names)
```

```python
code_list = pd.read_excel(data_guide_path, sheet_name="2024_code_list")
code_list.head(10)
```

```python
fields_to_check = ["local_authority_district","junction_control","age_of_vehicle","age_of_driver","engine_capacity_cc","casualty_imd_decile","enhanced_casualty_severity"]

selected_codes = code_list[code_list["field name"].isin(fields_to_check)]

display(selected_codes)
```

```python
minus_one_definitions = code_list[code_list["code/format"].astype(str).str.strip() == "-1"]

display(minus_one_definitions[["table", "field name", "code/format", "label", "note"]])
```


### Findings — Official Code Definitions

The official STATS19 data guide contains 63 entries explicitly defining the code `-1`.

The guide confirms that `-1` can represent different situations depending on the variable:

- **Data missing or out of range:** Information is unavailable or invalid.
- **Code deprecated:** A field is no longer used under the current reporting specification.
- **Historical compatibility:** A record predates the introduction of a particular field.

For example, `local_authority_district` is deprecated, explaining why all 2025 collision records contain `-1`.

Other variables, including `first_road_class`, `casualty_imd_decile`, and `casualty_distance_banding`, use `-1` to represent missing or out-of-range information.

These findings demonstrate that coded values must be interpreted individually using the official data dictionary.

**Conclusion:** The data-cleaning stage will distinguish missing, invalid, deprecated, and historically unavailable information. No coded values have been modified during this assessment.

### Investigation of Other Special Codes

This section identifies codes labelled as unknown, missing,not recorded, unavailable, or not applicable in the official STATS19 data dictionary.

```python
keywords = ("unknown|missing|not known|not recorded|""not applicable|unavailable|out of range|deprecated")

special_codes = code_list[code_list["label"].astype("string").str.contains(keywords, case=False, na=False, regex=True)]

print("Matching code definitions:", len(special_codes))

display(special_codes[["table", "field name", "code/format", "label", "note"]])
```


### Findings — Other Special Codes

The official STATS19 data dictionary contains 105 entries whose labels match terms associated with unknown, missing, unavailable, deprecated, or inapplicable information.

The analysis confirms that special codes vary across variables.

Examples include:
- `road_type = 9`: Unknown
- `first_road_number = -1`: Unknown
- `local_authority_district = -1`: Code deprecated
- `casualty_type = 99`: Unknown vehicle type (self-reported only)

Not all matching categories represent completely missing information. Some retain partial information and may be analytically meaningful.

**Conclusion:** Missing-value handling must be performed at the individual variable level using official definitions. Universal replacement of codes such as `9`, `99`, or `-1` would risk corrupting valid categorical information.

## Dataset Relationships and Integrity

### Duplicate and Identifier Analysis

This section examines duplicate records, identifier uniqueness,
and relationships between collision, vehicle, and casualty datasets.

```python
for name, df in datasets.items():
    print(f"\n{name} Dataset")
    print("-" * 45)

    print("Total records:", len(df))
    print("Exact duplicate rows:", df.duplicated().sum())
    print("Unique collision IDs:", df["collision_index"].nunique())
    print("Repeated collision IDs:", df["collision_index"].duplicated().sum())
```


### Findings — Duplicate and Identifier Analysis

The duplicate and identifier analysis produced the following results:

- **Collisions:** 101,525 records, 101,525 unique collision identifiers, and zero exact duplicate rows.
- **Vehicles:** 183,948 records, 101,525 unique collision identifiers, and zero exact duplicate rows.
- **Casualties:** 127,883 records, 101,525 unique collision identifiers, and zero exact duplicate rows.

The vehicle dataset contains 82,423 repeated collision identifiers, while the casualty dataset contains 26,358.

These repetitions are expected because a single collision may involve multiple vehicles and casualties.

The results support a one-to-many relationship between collisions and vehicles, and between collisions and casualties.

**Conclusion:** No exact duplicate rows were detected. Repeated collision identifiers in the vehicle and casualty datasets represent legitimate relational structure rather than duplicate observations. Further checks are required to verify referential integrity and record-level identifiers.

### Referential Integrity Verification

This section verifies whether vehicle and casualty records reference valid collisions in the collision dataset.

```python
collision_ids = set(collisions["collision_index"])
vehicle_ids = set(vehicles["collision_index"])
casualty_ids = set(casualties["collision_index"])

print("Vehicle IDs not found in collisions:",len(vehicle_ids - collision_ids))

print("Casualty IDs not found in collisions:",len(casualty_ids - collision_ids))

print("Collisions without vehicle records:",len(collision_ids - vehicle_ids))

print("Collisions without casualty records:",len(collision_ids - casualty_ids))
```


### Findings — Referential Integrity

The collision identifier comparison produced the following results:

- Vehicle collision IDs not found in the collision dataset: 0
- Casualty collision IDs not found in the collision dataset: 0
- Collisions without vehicle records: 0
- Collisions without casualty records: 0

All 101,525 collision identifiers are consistently represented across the collision, vehicle, and casualty datasets.

This confirms that the datasets share a complete set of collision-level identifiers and supports their integration using `collision_index`.

**Conclusion:** Collision-level referential integrity has been verified. Additional checks are required to validate vehicle and casualty reference numbers within individual collisions.

### Vehicle and Casualty Reference Validation

This section verifies the uniqueness of vehicle and casualty
reference numbers within each collision and investigates
relationships between casualty and vehicle records.

```python
vehicle_key_duplicates = vehicles.duplicated(subset=["collision_index", "vehicle_reference"]).sum()
casualty_key_duplicates = casualties.duplicated(subset=["collision_index", "casualty_reference"]).sum()

print("Duplicate vehicle composite keys:", vehicle_key_duplicates)
print("Duplicate casualty composite keys:", casualty_key_duplicates)
```

```python
vehicle_keys = pd.MultiIndex.from_frame(vehicles[["collision_index", "vehicle_reference"]])
casualty_vehicle_keys = pd.MultiIndex.from_frame(casualties[["collision_index", "vehicle_reference"]])
unmatched_vehicle_refs = ~casualty_vehicle_keys.isin(vehicle_keys)

print("Casualty records with unmatched vehicle references:",unmatched_vehicle_refs.sum())
print("Percentage unmatched:",round(unmatched_vehicle_refs.mean() * 100, 2),"%")
```


### Findings — Vehicle and Casualty Reference Validation

The composite-key and referential-integrity checks produced the following results:

- Duplicate vehicle composite keys: 0
- Duplicate casualty composite keys: 0
- Casualty records with unmatched vehicle references: 0
- Percentage of unmatched casualty-to-vehicle references: 0%

The combination of `collision_index` and `vehicle_reference` uniquely identifies each vehicle record.

Similarly, the combination of `collision_index` and `casualty_reference` uniquely identifies each casualty record.

All casualty records reference a valid vehicle record within the same collision.

**Conclusion:** The datasets demonstrate strong relational consistency at both collision and record levels. No duplicate composite keys or unmatched casualty-to-vehicle references were detected. These results support reliable dataset integration during the feature engineering stage.
